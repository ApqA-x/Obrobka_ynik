```python
import matplotlib.pyplot as plt
from pathlib import Path
from segment_objects import create_result, read_image, segment_objects

image_path = Path('images/IMG_20251006_103517.jpg')
image = read_image(image_path)
mask, objects = segment_objects(image)
result = create_result(image, mask, objects)

print(f'Кількість об’єктів: {objects}')
plt.figure(figsize=(12, 9))
plt.imshow(result[:, :, ::-1])
plt.axis('off')
plt.show()
```
